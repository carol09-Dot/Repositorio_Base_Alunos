```python
import numpy as np
from random import randint

notas = []
for i in range (100_000_000):
    notas.append(randint(0,100))

notas
```

```python
# Divide por 100 e multiplica por 2

notas_convertidas = []
for nota in notas:
    nova_nota = nota / 100 * 2
    notas_convertidas.append(nova_nota)

notas_convertidas 
```

```python
#Array

array_notas = np.array(notas) #Criando o array de notas
array_notas
```

```python
#É possível realizar as operações sem fazer for/while
array_notas = array_notas / 100 * 2
```

```python
#Podemos realizar realizar QUALQUER tipo de operação
array_notas

#todo array de verdadeiro ou falso pode ser usado como filtro
array_notas[array_notas <1]
```

```python
# Exercício 1

precos = np.array([20.0, 45.50, 120,0, 85.0, 10.0])

precos_atualizados = precos *1.15

print(precos_atualizados)
```

```python
#Exercício 2

mascara = precos_atualizados > 100
print(mascara)

precos_caros = precos_atualizados[mascara]
print(precos_caros)
```

```python
#Exercícios 3

idades = np.array([15, 18, 22, 14, 30, 17, 45])

maior_idade = idades [idades >= 18]
print (maior_idade)
```

```python
# Exercício 4
notas = np.array([5.5, 7.0, 4.5, 9.5, 8.0])

bonus_nota = notas + 0.5

print(bonus_nota)
```

```python
# Exercício 5
estoque = np.array([10, -2, 5, 0, -1, 20])

estoque_valido = estoque [estoque >=0]

print(estoque_valido)
```

```python
# Exercício 6
vendas = np.array([150, 200, 100, 350, 400])

total_venda = np.sum(vendas)
media_venda = np.mean(vendas)

print("Total vendido na semana:", total_venda )
print("Média diária de vendas:", media_venda)
```

```python
# Exercício 7
idades = np.array([12, 25, 30, 45, 15, 60])

idades_maiores = idades[(idades > 20) & (idades < 40)]

print(idades_maiores)
```

```python
# Exercício 8
temp = np.array([25, -999, 30, 22, -999])

temp_corrigida = np.where(temp == -999, 0, temp)

print(temp_corrigida)
```

```python
# Exercício 9
valores = np.array([1.5, 2.0, np.nan, 4.5, np.nan])

mascara = np.isnan(valores)

valores_validos = valores [~mascara]
print(valores_validos)
```
